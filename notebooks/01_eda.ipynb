```python
import pandas as pd
from ucl.db.session import engine

matches = pd.read_sql("SELECT * FROM matches", engine, parse_dates=["utc_date"])
teams = pd.read_sql("SELECT * FROM teams", engine)

print(matches.groupby("season").size())            # matches per season
print(matches.status.value_counts())               # FINISHED vs SCHEDULED
print(matches.isna().sum())                        # missing values
print(matches.duplicated("id").sum())              # should be 0
finished = matches[matches.status == "FINISHED"]
print(finished.assign(
    result=lambda d: (d.home_goals > d.away_goals).map({True: "H", False: "other"})
).result.value_counts(normalize=True))             # home win rate
```
